```python
from pypokerengine.api.game import setup_config, start_poker
from pypokerengine.players import BasePokerPlayer
import random

#Unifinished - Will likely not use a completly random player
class SimpleRandomPlayer(BasePokerPlayer):
    def declare_action(self, valid_actions, hole_card, round_state):
        #Decide action at random
        choice = random.choice(valid_actions)
        action = choice['action']
        if action == 'raise':
            ammount = choice['amount']['min']
        else:
            ammount = choice['amount']
        return action, ammount

    def receive_game_start_message(self, game_info):
        pass
    def receive_round_start_message(self, round_count, hole_card, seats):
        pass
    def receive_street_start_message(self, street, round_state):
        pass
    def receive_game_update_message(self, action, round_state):
        pass
    def receive_round_result_message(self, winners, hand_info, round_state):
        pass

config = setup_config(max_round=1, initial_stack=1000, small_blind_amount=10)

main = BehavioralAnalyzerPlayer()
config.register_player(name="Main", algorithm=main)
config.register_player(name="Rand", algorithm=SimpleRandomPlayer())
config.register_player(name="Rand2", algorithm=SimpleRandomPlayer())

start_poker(config)

#Will run numerous trials to detemine weights

#start_poker(config, verbose=0)


#print("\nTracked Opponent Betting Habits:")
#for uuid, actions in analyzer.opponent_histories.items():
#    print(f"Opponent UUID: {uuid}")
#    for act, count in actions.items():
#        print(f"  - Action: {act:<6} | Count: {count}")
```
